# How do you find the minimum value in each column of a DataFrame?

**Question:** How do you find the minimum value in each column of a DataFrame?



---



**Introduction:**



Finding the minimum value in each column of a DataFrame is a common operation in data analysis. It allows us to quickly identify the smallest value in each individual column, providing insights into the range and distribution of data. In Pandas, you can use the `min()` function to accomplish this task efficiently.



**Finding the Minimum Value in Each Column:**



To find the minimum value in each column of a DataFrame, we can use the `min()` function in Pandas. This function calculates the minimum value for each column separately and returns the result as a Series object, where the index represents the column names.



**Example:**



Let's demonstrate how to find the minimum value in each column of a DataFrame using the Titanic dataset.

```

import pandas as pd



# Load the Titanic dataset into a DataFrame

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Find the minimum value in each column

min_values = titanic_df.min()



print(min_values)
```

**Explanation:**



- We start by importing the Pandas library with the alias `pd`.

- Next, we load the Titanic dataset from the provided URL into a Pandas DataFrame `titanic_df` using the `pd.read_csv()` function.

- To find the minimum value in each column, we apply the `min()` function directly to the DataFrame `titanic_df`.

- The `min()` function calculates the minimum value for each column separately and returns the result as a Series object `min_values`, where the index represents the column names.

- Finally, we print the Series `min_values` to display the minimum value in each column of the DataFrame.



**Conclusion:**



Finding the minimum value in each column of a DataFrame is a straightforward operation in Pandas using the `min()` function. By applying this function to the DataFrame, you can quickly identify the smallest value in each individual column, which is useful for understanding the distribution and range of data within the dataset.
