# How do you handle NaN values in a DataFrame?

**Question:** How do you handle NaN values in a DataFrame?



---



**Introduction:**



In real-world datasets, missing values, often represented as NaN (Not a Number), are a common occurrence. These missing values can arise due to various reasons such as data collection errors, equipment malfunctions, or simply because the information was not available. It's essential to handle these missing values appropriately to ensure accurate analysis and modeling. In this tutorial, we'll explore different methods to handle NaN values in a DataFrame using Pandas.



**Handling NaN Values:**



Pandas provides several methods for handling NaN values in a DataFrame:



1. **Detecting NaN Values:**

   - `isna()`: Returns a DataFrame of the same shape as the original DataFrame but with Boolean values indicating whether each element is NaN.

   - `notna()`: Returns the opposite of `isna()`, indicating non-missing values.



2. **Handling NaN Values:**

   - `dropna()`: Drops rows or columns containing NaN values.

   - `fillna()`: Fills NaN values with a specified value or method.



**Example:**



Let's demonstrate how to handle NaN values in a DataFrame using the Titanic dataset. We'll load the dataset, introduce some NaN values, and then use the mentioned methods to handle them.

````

import pandas as pd



# Load Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Introduce NaN values

titanic_df.loc[2, 'Age'] = None  # Setting a specific value to NaN



# Detect NaN values

print("NaN values before handling:")

print(titanic_df.isna().sum())



# Handling NaN values

# Drop rows with NaN values in 'Age' column

titanic_df.dropna(subset=['Age'], inplace=True)



# Fill NaN values in 'Cabin' column with 'Unknown'

titanic_df['Cabin'].fillna('Unknown', inplace=True)



# Verify NaN values after handling

print("\nNaN values after handling:")

print(titanic_df.isna().sum())
````
