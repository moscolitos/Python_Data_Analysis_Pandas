# How do you create a stacked bar plot from a DataFrame?

**Question:** How do you create a stacked bar plot from a DataFrame?



---



**Introduction:**



A stacked bar plot is a useful visualization technique for comparing the distribution of categorical variables across different groups. In Pandas, you can easily create a stacked bar plot from a DataFrame using the `plot()` function with the `kind='bar'` parameter and setting `stacked=True`.



**Creating a Stacked Bar Plot:**



To create a stacked bar plot from a DataFrame in Pandas, you can use the `plot()` function with the `kind='bar'` parameter and set `stacked=True`. This function generates a stacked bar plot where each category is represented as a separate stack, and the height of each stack corresponds to the frequency or proportion of data points in that category.



**Example:**



Let's illustrate how to create a stacked bar plot from the Titanic dataset. In this example, we'll create a stacked bar plot to visualize the distribution of survival status ('Survived') across different passenger classes ('Pclass').

```

import pandas as pd

import matplotlib.pyplot as plt



# Load the Titanic dataset into a DataFrame

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Create a stacked bar plot of 'Survived' across 'Pclass'

stacked_bar_plot = titanic_df.groupby(['Pclass', 'Survived']).size().unstack().plot(kind='bar', stacked=True)



# Set plot labels and title

plt.xlabel('Passenger Class')

plt.ylabel('Count')

plt.title('Survival Status by Passenger Class')

plt.legend(title='Survived', labels=['No', 'Yes'])



# Show the plot

plt.show()
```

**Explanation:**



- First, we import the Pandas library with the alias `pd` and the Matplotlib library with the alias `plt`.

- Next, we load the Titanic dataset from the provided URL into a DataFrame named `titanic_df`.

- We use the `groupby()` function to group the data by 'Pclass' and 'Survived' columns and calculate the size of each group. Then, we use the `unstack()` function to pivot the data, converting it from long to wide format.

- Using the `plot()` function with `kind='bar'` and `stacked=True`, we create a stacked bar plot where each bar represents a passenger class ('Pclass') and the height of each stack represents the count of passengers who survived ('Survived=Yes') or did not survive ('Survived=No').

- Finally, we customize the plot by setting labels for the x-axis and y-axis, adding a title, and specifying the legend labels.



**Conclusion:**



Creating a stacked bar plot from a DataFrame in Pandas allows you to visually compare the distribution of categorical variables across different groups. By following the steps outlined in this tutorial, you can easily generate informative stacked bar plots to analyze and interpret the relationship between categorical variables in your dataset.
