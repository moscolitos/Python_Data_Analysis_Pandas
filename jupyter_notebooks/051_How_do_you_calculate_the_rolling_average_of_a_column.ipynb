# How do you calculate the rolling average of a column?

**Question:** How do you calculate the rolling average of a column?



---



**Introduction:**



In data analysis, especially time-series data, calculating the rolling average of a column is a common operation. A rolling average, also known as a moving average, smooths out fluctuations in data by averaging values in a specified window. Pandas provides a convenient method to compute rolling statistics, including rolling averages, using the `rolling()` function.



**Calculating Rolling Average with Pandas:**



To calculate the rolling average of a column in a DataFrame, you can use the `rolling()` function followed by the `mean()` function. The `rolling()` function creates a rolling window object, and the `mean()` function computes the mean value within each window.



**Example:**



Let's illustrate how to calculate the rolling average of the 'Fare' column in the Titanic dataset using Pandas.

```

import pandas as pd



# Load the Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Convert 'Fare' column to numeric (if not already)

titanic_df['Fare'] = pd.to_numeric(titanic_df['Fare'], errors='coerce')



# Calculate rolling average of 'Fare' column with window size 5

rolling_avg_fare = titanic_df['Fare'].rolling(window=5).mean()



# Print the result

print("Rolling Average of Fare (Window Size = 5):")

print(rolling_avg_fare)
```

**Explanation:**



- We start by importing the Pandas library with the alias `pd`.

- Next, we load the Titanic dataset from the provided URL using the `read_csv()` function from Pandas and store it in the DataFrame `titanic_df`.

- Since the 'Fare' column may contain non-numeric values, we convert it to numeric using the `pd.to_numeric()` function. The `errors='coerce'` parameter ensures that non-numeric values are converted to NaN (Not a Number).

- To calculate the rolling average of the 'Fare' column, we use the `rolling()` function on the 'Fare' column of the DataFrame `titanic_df`. We specify the window size as 5 using the `window` parameter.

- Finally, we compute the mean of each rolling window using the `mean()` function and store the result in the variable `rolling_avg_fare`.



**Conclusion:**



Calculating the rolling average of a column is a valuable technique in data analysis for identifying trends and patterns in time-series or sequential data. By leveraging the `rolling()` function in Pandas, users can easily compute rolling statistics, such as rolling averages, with customizable window sizes. This facilitates the extraction of meaningful insights from data and aids in making informed decisions based on historical trends.
