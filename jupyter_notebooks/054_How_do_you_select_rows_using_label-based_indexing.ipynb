# How do you select rows using label-based indexing?

**Question:** How do you select rows using label-based indexing?



---



**Introduction:**



In pandas, label-based indexing allows you to select rows from a DataFrame based on their index labels. This indexing method is useful when you want to retrieve specific rows using their index values or labels. Label-based indexing provides a flexible and intuitive way to access data, especially when the index contains meaningful labels or identifiers.



**Selecting Rows Using Label-Based Indexing:**



To select rows using label-based indexing in pandas, you can use the `.loc[]` accessor. This accessor allows you to specify the desired index labels or ranges and retrieve the corresponding rows from the DataFrame. You can provide a single label, a list of labels, or a range of labels to select rows based on their index.



**Example:**



Let's demonstrate how to select rows using label-based indexing with the Titanic dataset.

```

import pandas as pd



# Load the Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Set 'PassengerId' column as the index

titanic_df.set_index('PassengerId', inplace=True)



# Select rows with index labels 1 and 2

selected_rows = titanic_df.loc[[1, 2]]



# Display the selected rows

print(selected_rows)
```

**Explanation:**



- We start by importing the pandas library with the alias `pd`.

- Next, we load the Titanic dataset from the provided URL using the `read_csv()` function and store it in the DataFrame `titanic_df`.

- To use label-based indexing, we set the 'PassengerId' column as the index of the DataFrame using the `set_index()` function with the `inplace=True` parameter, which modifies the DataFrame in place.

- We then use the `.loc[]` accessor to select rows with index labels 1 and 2 by passing a list `[1, 2]` inside the `.loc[]` accessor.

- Finally, we display the selected rows using the `print()` function.



**Conclusion:**



Label-based indexing in pandas provides a convenient way to select rows from a DataFrame based on their index labels. By using the `.loc[]` accessor, users can specify the desired index labels or ranges and retrieve the corresponding rows effectively. Whether you need to retrieve specific rows or subsets of data based on their index labels, label-based indexing offers a versatile solution for data selection in pandas.
