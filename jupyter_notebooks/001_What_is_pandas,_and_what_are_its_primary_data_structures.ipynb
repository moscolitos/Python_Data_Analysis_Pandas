# What is pandas, and what are its primary data structures?

**Question:** What is pandas, and what are its primary data structures?



---



**Introduction:**



Pandas is a powerful and popular open-source Python library used for data manipulation and analysis. It provides easy-to-use data structures and functions to work with structured data. In this tutorial, we'll delve into the fundamentals of pandas, including its primary data structures.



**Primary Data Structures in Pandas:**



Pandas primarily revolves around two main data structures:



1. **Series:** A one-dimensional labeled array capable of holding any data type (integers, strings, floating-point numbers, Python objects, etc.). It is similar to a one-dimensional array or list in Python but with additional functionalities. Each element in a Series has a label associated with it, which is called the index.



2. **DataFrame:** A two-dimensional labeled data structure with columns of potentially different types. It is akin to a spreadsheet or SQL table, where data is organized into rows and columns. Each column in a DataFrame is a Series. DataFrames allow you to store and manipulate heterogeneous tabular data effectively.



**Exploring Series and DataFrame:**



Let's dive into each of these data structures with examples using the Titanic dataset.

```

import pandas as pd



# Reading the Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_data = pd.read_csv(url)



# Creating a Series from a list

s = pd.Series([1, 3, 5, 7, 9])

print("Series:")

print(s)
```

In the above code:

- We imported pandas as `pd`, following the conventional alias.

- Loaded the Titanic dataset using `pd.read_csv()` function. We provided the URL of the dataset.

- Created a Series `s` using `pd.Series()` constructor with a list of integers.



Now, let's explore DataFrame:

```

# Creating a DataFrame from a dictionary

data = {

    'Name': ['John', 'Anna', 'Peter', 'Linda'],

    'Age': [25, 35, 30, 28],

    'Gender': ['Male', 'Female', 'Male', 'Female']

}

df = pd.DataFrame(data)

print("\nDataFrame:")

print(df)
```

In the above code:

- We created a dictionary `data` containing columns 'Name', 'Age', and 'Gender'.

- Used `pd.DataFrame()` constructor to create a DataFrame `df` from the dictionary.



**Conclusion:**



Pandas provides versatile data structures, Series and DataFrame, which form the backbone of data manipulation and analysis in Python.
