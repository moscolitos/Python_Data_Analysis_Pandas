# How do you find the maximum value in each column of a DataFrame?

**Question:** How do you find the maximum value in each column of a DataFrame?



---



**Introduction:**



Finding the maximum value in each column of a DataFrame is a common operation in data analysis. It helps identify the highest value present in each column, which can be useful for understanding the range and distribution of data. In Pandas, you can use the `max()` function to compute the maximum value in each column of a DataFrame.



**Finding the Maximum Value in Each Column:**



To find the maximum value in each column of a DataFrame, you can use the `max()` function in Pandas. This function computes the maximum value along a specified axis, which can be either rows (axis=0) or columns (axis=1). By default, the `max()` function operates along columns (axis=0), so you can directly apply it to the DataFrame to find the maximum value in each column.



**Example:**



Let's demonstrate how to find the maximum value in each column of a DataFrame using the Titanic dataset. In this example, we'll compute the maximum value in each column of the DataFrame.

```

import pandas as pd



# Load the Titanic dataset into a DataFrame

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Find the maximum value in each column

max_values = titanic_df.max()



print("Maximum values in each column:")

print(max_values)
```

**Explanation:**



- We start by importing the Pandas library with the alias `pd`.

- Next, we load the Titanic dataset from the provided URL into a Pandas DataFrame `titanic_df` using the `pd.read_csv()` function.

- To find the maximum value in each column, we apply the `max()` function directly to the DataFrame `titanic_df`. This computes the maximum value along each column (axis=0) by default.

- Finally, we print the maximum values in each column using the `print()` function.



**Conclusion:**



Finding the maximum value in each column of a DataFrame is essential for understanding the distribution and range of data. By using the `max()` function in Pandas, you can easily compute the maximum value in each column of a DataFrame. This allows you to gain insights into the data and identify the highest values present in each column.
