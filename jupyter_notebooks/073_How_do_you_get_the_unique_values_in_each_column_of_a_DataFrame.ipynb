# How do you get the unique values in each column of a DataFrame?

**Question:** How do you get the unique values in each column of a DataFrame?



---



**Introduction:**



In data analysis, it's often essential to understand the unique values present in each column of a DataFrame. Pandas provides a convenient method, `unique()`, to extract the unique values from a DataFrame column. This function returns an array of unique elements in the specified column.



**Getting Unique Values:**



To get the unique values in each column of a DataFrame, you can iterate over the columns and apply the `unique()` function to each column individually. This allows you to obtain a list of unique values for each column in the DataFrame.



**Example:**



Let's demonstrate how to get the unique values in each column of the Titanic dataset. We'll iterate over the columns of the DataFrame and print the unique values for each column.

```

import pandas as pd



# Load the Titanic dataset into a DataFrame

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Iterate over the columns of the DataFrame

for column in titanic_df.columns:

    # Get the unique values in the current column

    unique_values = titanic_df[column].unique()

    

    # Print the column name and its unique values

    print(f"Unique values in column '{column}':")

    print(unique_values)

    print()
```

**Explanation:**



- We start by importing the Pandas library with the alias `pd`.

- Next, we load the Titanic dataset from the provided URL into a DataFrame named `titanic_df`.

- We then iterate over the columns of the DataFrame using a for loop.

- Within the loop, we use the `unique()` function to get the unique values in the current column.

- Finally, we print the column name and its corresponding unique values.



**Conclusion:**



Obtaining the unique values in each column of a DataFrame is crucial for data exploration and analysis. By using the `unique()` function in Pandas, you can efficiently extract the unique elements present in each column. This tutorial provides a step-by-step example of how to achieve this task using the Titanic dataset.
