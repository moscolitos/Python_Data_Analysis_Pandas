# How do you append a new row to a DataFrame?

**Question:** How do you append a new row to a DataFrame?



---



**Introduction:**



In data analysis, it's common to need to add new data to an existing DataFrame. This could be a single row of data that needs to be appended to the DataFrame. Pandas provides several methods for adding new rows to a DataFrame, allowing you to efficiently update your dataset with additional information.



**Appending a New Row to a DataFrame:**



To append a new row to a DataFrame in Pandas, you can use the `append()` method. This method allows you to concatenate two DataFrames along either axis, effectively adding the rows of one DataFrame to the end of the other.



**Example:**



Let's demonstrate how to append a new row to a DataFrame using the Titanic dataset.

```

import pandas as pd



# Load the Titanic dataset into a DataFrame

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Create a dictionary representing the new row to be appended

new_row = {'PassengerId': 1000, 'Survived': 1, 'Pclass': 1, 'Name': 'Smith, Miss. Jane', 

           'Sex': 'female', 'Age': 25, 'SibSp': 0, 'Parch': 0, 'Ticket': 'PC 17582', 

           'Fare': 100, 'Cabin': 'C85', 'Embarked': 'C'}



# Convert the dictionary to a DataFrame

new_row_df = pd.DataFrame([new_row])



# Append the new row to the original DataFrame

titanic_df = titanic_df.append(new_row_df, ignore_index=True)



print(titanic_df.tail())
```

**Explanation:**



- We begin by importing the Pandas library with the alias `pd`.

- Next, we load the Titanic dataset from the provided URL into a Pandas DataFrame `titanic_df` using the `pd.read_csv()` function.

- We then create a dictionary `new_row` representing the new row of data that we want to append to the DataFrame.

- Using the `pd.DataFrame()` constructor, we convert the dictionary `new_row` into a DataFrame `new_row_df`.

- Finally, we append the new row DataFrame `new_row_df` to the original DataFrame `titanic_df` using the `append()` method. We set the `ignore_index` parameter to `True` to reset the index of the resulting DataFrame.



**Conclusion:**



Appending a new row to a DataFrame in Pandas allows you to dynamically update your dataset with additional data. By using the `append()` method, you can efficiently concatenate DataFrames and add new rows, enabling you to expand your dataset as needed during the data analysis process.
