# How do you save a DataFrame to a JSON file?

**Question:** How do you save a DataFrame to a JSON file?



---



**Introduction:**



Once you've performed data analysis and manipulation on a DataFrame in Pandas, you may want to save the results for future reference or to share them with others. One common format for saving structured data is JSON (JavaScript Object Notation), which is human-readable and widely supported by various programming languages and applications. In this tutorial, we'll explore how to save a Pandas DataFrame to a JSON file using Python.



**Saving DataFrame to JSON:**



Pandas provides a method called `to_json()` that allows you to save a DataFrame to a JSON file. This method offers several parameters to customize the output JSON format, including the orientation, compression, and whether to include index or column labels.



**Example:**



Let's demonstrate how to use the `to_json()` method to save a DataFrame to a JSON file.

```

import pandas as pd



# Load Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Save DataFrame to JSON file

json_filename = "titanic_data.json"

titanic_df.to_json(json_filename, orient="records", lines=True)



print("DataFrame saved to", json_filename)
```

**Explanation:**



- We begin by loading the Titanic dataset into a DataFrame `titanic_df` using the `read_csv()` function from Pandas.

- Next, we specify the filename for the JSON file we want to save the DataFrame to, in this case, "titanic_data.json".

- We then use the `to_json()` method on the DataFrame `titanic_df` to save it to the specified JSON file. In this example:

  - The `orient` parameter is set to "records", which specifies that each row of the DataFrame will be represented as a JSON object.

  - The `lines` parameter is set to True, indicating that each JSON object will be written on a separate line in the output file.

- Finally, we print a message confirming that the DataFrame has been successfully saved to the JSON file.



**Conclusion:**



Saving a Pandas DataFrame to a JSON file is straightforward using the `to_json()` method. This tutorial provided a clear example of how to accomplish this task, demonstrating the use of parameters to customize the output JSON format. By following the steps outlined in this tutorial, you can easily save your DataFrame to a JSON file and preserve your data for future use or sharing.
