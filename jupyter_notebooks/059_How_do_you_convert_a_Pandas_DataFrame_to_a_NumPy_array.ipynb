# How do you convert a Pandas DataFrame to a NumPy array?

**Question:** How do you convert a Pandas DataFrame to a NumPy array?



---



**Introduction:**



Pandas and NumPy are two powerful libraries widely used in data science and machine learning tasks. While Pandas provides data structures like DataFrames for handling tabular data, NumPy offers support for numerical computing with its ndarray (n-dimensional array) data structure. In some cases, you may need to convert a Pandas DataFrame to a NumPy array to leverage the functionalities offered by NumPy or to interface with other libraries that expect NumPy arrays as input.



**Converting a Pandas DataFrame to a NumPy Array:**



To convert a Pandas DataFrame to a NumPy array, you can use the `.values` attribute of the DataFrame. The `.values` attribute returns the underlying data as a NumPy array. This conversion allows you to perform various array operations and computations using NumPy's extensive functionality.



**Example:**



Let's demonstrate how to convert a Pandas DataFrame to a NumPy array using the Titanic dataset.

```

import pandas as pd



# Load the Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Convert the DataFrame to a NumPy array

numpy_array = titanic_df.values



# Display the first few rows of the NumPy array

print(numpy_array[:5])
```

**Explanation:**



- We begin by importing the pandas library with the alias `pd`.

- Next, we load the Titanic dataset from the provided URL using the `read_csv()` function and store it in the DataFrame `titanic_df`.

- To convert the DataFrame to a NumPy array, we simply access the `.values` attribute of the DataFrame and assign it to the variable `numpy_array`.

- Finally, we display the first few rows of the NumPy array using array slicing and the `print()` function.



**Conclusion:**



Converting a Pandas DataFrame to a NumPy array is straightforward and can be accomplished using the `.values` attribute of the DataFrame. This conversion enables seamless integration with NumPy's extensive array manipulation and computation capabilities. Whether you need to perform numerical computations, array operations, or interface with other libraries, converting Pandas DataFrame to NumPy arrays provides flexibility and interoperability in your data analysis and machine learning workflows.
