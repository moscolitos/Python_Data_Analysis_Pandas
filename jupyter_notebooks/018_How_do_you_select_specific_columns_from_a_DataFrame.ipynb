# How do you select specific columns from a DataFrame?

**Question:** How do you select specific columns from a DataFrame?



---



**Introduction:**



In data analysis, it's common to work with large datasets containing numerous columns. While analyzing such datasets, it's often necessary to select specific columns of interest for further analysis or visualization. Pandas provides intuitive methods to accomplish this task efficiently, allowing analysts to retrieve and work with specific columns from a DataFrame. Understanding how to select specific columns from a DataFrame is essential for various data manipulation and analysis tasks. In this tutorial, we'll explore how to select specific columns from a DataFrame, providing detailed explanations and coding examples.



**Selecting Specific Columns from a DataFrame:**



Pandas offers several methods to select specific columns from a DataFrame:

1. **Using bracket notation**: This method allows selecting one or more columns by specifying their names inside square brackets `[]`.

2. **Using `.loc[]`**: This method is primarily label-based and allows selecting columns by their column labels.

3. **Using `.iloc[]`**: This method is primarily integer-based and allows selecting columns by their integer index positions.



**Example:**



Let's demonstrate how to select specific columns from the Titanic dataset DataFrame:

```

import pandas as pd



# URL of the Titanic dataset CSV file

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"



# Reading CSV file into DataFrame

titanic_df = pd.read_csv(url)



# Using bracket notation to select specific columns

selected_columns_1 = titanic_df[['Name', 'Sex', 'Age']]



# Using .loc[] to select specific columns

selected_columns_2 = titanic_df.loc[:, ['Survived', 'Pclass', 'Fare']]



# Using .iloc[] to select specific columns

selected_columns_3 = titanic_df.iloc[:, [1, 3, 5]]



# Displaying the selected columns

print("Selected columns using bracket notation:")

print(selected_columns_1.head())

print("\nSelected columns using .loc[]:")

print(selected_columns_2.head())

print("\nSelected columns using .iloc[]:")

print(selected_columns_3.head())
```

In the above code:

- We imported pandas as `pd`, following the conventional alias.

- Specified the URL of the Titanic dataset CSV file.

- Used the `pd.read_csv()` function to read the CSV file from the provided URL and create a DataFrame `titanic_df`.

- Demonstrated three different methods to select specific columns:

  - Using bracket notation `[['Name', 'Sex', 'Age']]` to select columns 'Name', 'Sex', and 'Age'.

  - Using `.loc[]` with `[:, ['Survived', 'Pclass', 'Fare']]` to select columns 'Survived', 'Pclass', and 'Fare'.

  - Using `.iloc[]` with `[:, [1, 3, 5]]` to select columns at integer index positions 1, 3, and 5.

- Displayed the selected columns using the `.head()` method to show the first few rows.
