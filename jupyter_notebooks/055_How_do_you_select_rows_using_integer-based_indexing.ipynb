# How do you select rows using integer-based indexing?

**Question:** How do you select rows using integer-based indexing?



---



**Introduction:**



In pandas, integer-based indexing allows you to select rows from a DataFrame based on their positional index. This indexing method is useful when you want to retrieve specific rows using their integer positions rather than their index labels. Integer-based indexing provides a straightforward way to access rows, especially when the index does not contain meaningful labels or identifiers.



**Selecting Rows Using Integer-Based Indexing:**



To select rows using integer-based indexing in pandas, you can use the `.iloc[]` accessor. This accessor allows you to specify the desired integer positions or ranges and retrieve the corresponding rows from the DataFrame. You can provide a single integer, a list of integers, or a range of integers to select rows based on their positions.



**Example:**



Let's demonstrate how to select rows using integer-based indexing with the Titanic dataset.

```

import pandas as pd



# Load the Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Select rows with integer positions 0 to 4

selected_rows = titanic_df.iloc[0:5]



# Display the selected rows

print(selected_rows)
```

**Explanation:**



- We start by importing the pandas library with the alias `pd`.

- Next, we load the Titanic dataset from the provided URL using the `read_csv()` function and store it in the DataFrame `titanic_df`.

- To use integer-based indexing, we can use the `.iloc[]` accessor. In this example, we select rows with integer positions 0 to 4 by passing the range `0:5` inside the `.iloc[]` accessor.

- Finally, we display the selected rows using the `print()` function.



**Conclusion:**



Integer-based indexing in pandas provides a simple and efficient way to select rows from a DataFrame based on their positional index. By using the `.iloc[]` accessor, users can specify the desired integer positions or ranges and retrieve the corresponding rows effectively. Whether you need to retrieve specific rows or subsets of data based on their integer positions, integer-based indexing offers a reliable solution for data selection in pandas.
