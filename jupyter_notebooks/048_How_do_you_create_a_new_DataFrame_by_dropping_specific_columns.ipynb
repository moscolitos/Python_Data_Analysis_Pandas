# How do you create a new DataFrame by dropping specific columns?

**Question:** How do you create a new DataFrame by dropping specific columns?



---



**Introduction:**



In data analysis, it's common to work with large datasets containing numerous columns, some of which may not be relevant to the analysis or modeling process. In such cases, it's necessary to create a new DataFrame with only the essential columns. Pandas provides a convenient method to accomplish this task: the `drop()` function.



**Creating a New DataFrame by Dropping Specific Columns with Pandas:**



The `drop()` function in Pandas allows you to remove specific columns from a DataFrame and create a new DataFrame with the remaining columns. This function provides flexibility by allowing you to specify columns to drop either by their labels or their index. Additionally, you can control whether the operation is performed inplace or a new DataFrame is returned.



**Example:**



Let's illustrate how to create a new DataFrame by dropping specific columns from the Titanic dataset.

```

import pandas as pd



# Load the Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

df = pd.read_csv(url)



# Drop specific columns and create a new DataFrame

columns_to_drop = ['SibSp', 'Parch', 'Ticket', 'Cabin', 'Embarked']

new_df = df.drop(columns=columns_to_drop)



# Display the first few rows of the new DataFrame

print(new_df.head())
```

**Explanation:**



- We begin by importing the Pandas library with the alias `pd`.

- Next, we load the Titanic dataset into a DataFrame `df` using the `read_csv()` function from Pandas. We specify the URL of the dataset as the input.

- To create a new DataFrame by dropping specific columns, we use the `drop()` method on the original DataFrame `df`. We pass the `columns` parameter and provide a list of column labels to be dropped.

- The `drop()` method returns a new DataFrame with the specified columns removed, which we assign to the variable `new_df`.

- Finally, we display the first few rows of the new DataFrame using the `head()` method to verify that the specified columns have been dropped.



**Conclusion:**



The `drop()` function in Pandas provides a straightforward way to create a new DataFrame by removing specific columns from an existing DataFrame. By following the example and explanations provided in this tutorial, users can effectively manipulate their datasets and extract only the relevant information for their analysis or modeling tasks.
