# How do you get descriptive statistics for each column in a DataFrame?

**Question:** How do you get descriptive statistics for each column in a DataFrame?



---



**Introduction:**



Descriptive statistics provide a summary of the main characteristics of a dataset. In data analysis, understanding these statistics is crucial for gaining insights into the distribution, central tendency, and variability of the data. Pandas, a powerful data manipulation library in Python, offers convenient functions to compute descriptive statistics for each column in a DataFrame.



**Getting Descriptive Statistics with Pandas:**



Pandas provides the `describe()` method, which generates descriptive statistics for numeric columns in a DataFrame. These statistics include count, mean, standard deviation, minimum, maximum, and percentiles. Additionally, you can use specific methods like `mean()`, `median()`, `std()`, `min()`, `max()`, and `quantile()` to compute individual statistics for each column.



**Example:**



Let's illustrate how to get descriptive statistics for each column in a DataFrame using the Titanic dataset.

```

import pandas as pd



# Load the Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

df = pd.read_csv(url)



# Display descriptive statistics for numeric columns

print(df.describe())



# Calculate mean and median for specific columns

mean_age = df['Age'].mean()

median_fare = df['Fare'].median()



print("\nMean age:", mean_age)

print("Median fare:", median_fare)
```

**Explanation:**



- We start by importing the Pandas library with the alias `pd`.

- Next, we load the Titanic dataset into a DataFrame `df` using the `read_csv()` function from Pandas. We specify the URL of the dataset as the input.

- To get descriptive statistics for numeric columns in the DataFrame, we use the `describe()` method. This method computes count, mean, standard deviation, minimum, maximum, and percentiles for each numeric column.

- Additionally, we demonstrate how to calculate specific statistics for individual columns. We compute the mean age and median fare using the `mean()` and `median()` methods, respectively.

- Finally, we print the descriptive statistics for all numeric columns and the calculated mean age and median fare.



**Conclusion:**



Descriptive statistics play a crucial role in understanding the characteristics of a dataset. With Pandas, users can easily compute these statistics for each column in a DataFrame using the `describe()` method. Furthermore, specific statistics such as mean, median, standard deviation, minimum, and maximum can be calculated individually for selected columns. By following the example and explanations provided in this tutorial, users can effectively analyze and interpret the descriptive statistics of their datasets using Pandas.
