# How do you add a multi-level index to a DataFrame?

**Question:** How do you add a multi-level index to a DataFrame?



---



**Introduction:**



In pandas, an index serves as a unique identifier for rows in a DataFrame. It provides a way to access, manipulate, and analyze data efficiently. While a single-level index is suitable for many scenarios, there are cases where a multi-level index, also known as a hierarchical index, offers more flexibility and organization. A multi-level index consists of two or more levels of indexing, enabling users to represent and structure data in a more hierarchical manner.



**Adding a Multi-Level Index to a DataFrame:**



To add a multi-level index to a DataFrame in pandas, you can use the `set_index()` function. This function allows you to specify one or more columns as the index or create a hierarchical index by passing a list of column names. By defining multiple levels of indexing, you can organize the data hierarchically and perform advanced operations such as groupings, slicing, and reshaping more effectively.



**Example:**



Let's demonstrate how to add a multi-level index to the Titanic dataset based on the 'Pclass' and 'Sex' columns.

```

import pandas as pd



# Load the Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Set 'Pclass' and 'Sex' columns as the multi-level index

multi_index_df = titanic_df.set_index(['Pclass', 'Sex'])



# Display the DataFrame with the multi-level index

print(multi_index_df)
```

**Explanation:**



- We start by importing the pandas library with the alias `pd`.

- Next, we load the Titanic dataset from the provided URL using the `read_csv()` function and store it in the DataFrame `titanic_df`.

- To create a multi-level index, we use the `set_index()` function and pass a list of column names ('Pclass' and 'Sex') as arguments. This specifies that the DataFrame should be indexed hierarchically based on these columns.

- Finally, we display the DataFrame `multi_index_df` with the newly added multi-level index.



**Conclusion:**



Adding a multi-level index to a DataFrame in pandas allows for more structured and organized representation of data, particularly when dealing with complex datasets. By leveraging hierarchical indexing, users can perform advanced data manipulations and analysis more efficiently. With the `set_index()` function, pandas provides a simple yet powerful way to create multi-level indexes and unlock the full potential of hierarchical data organization.
