# How do you read data from a SQL database into a DataFrame?

**Question:** How do you read data from a SQL database into a DataFrame?



---



**Introduction:**



Data often resides in SQL databases, and you may need to analyze this data using Pandas in Python. Fortunately, Pandas provides convenient functions to read data from SQL databases directly into DataFrames, allowing for seamless integration between SQL and Python for data analysis tasks. In this tutorial, we'll explore how to read data from a SQL database into a Pandas DataFrame using Python.



**Reading Data from SQL Database:**



Pandas provides a function called `read_sql()` that allows you to execute a SQL query against a database and read the results into a DataFrame. This function requires a database connection and a SQL query as input parameters.



**Example:**



Let's demonstrate how to use the `read_sql()` function to read data from a SQL database into a DataFrame.

```

import pandas as pd

import sqlite3



# Create a connection to the SQLite database

conn = sqlite3.connect('titanic.db')



# Define the SQL query

sql_query = "SELECT * FROM titanic_data;"



# Read data from the database into a DataFrame

titanic_df = pd.read_sql(sql_query, conn)



# Close the database connection

conn.close()



# Display the first few rows of the DataFrame

print(titanic_df.head())
```

**Explanation:**



- We start by importing the necessary libraries: Pandas for data manipulation and SQLite3 for database connection.

- Next, we establish a connection to the SQLite database by calling `sqlite3.connect()` and passing the database filename ('titanic.db') as a parameter. You would replace this with the connection details of your specific SQL database.

- We define the SQL query that we want to execute against the database. In this example, we're selecting all columns from a table named 'titanic_data'.

- Using the `read_sql()` function from Pandas, we execute the SQL query and read the results into a DataFrame `titanic_df`. We pass the SQL query and the database connection as parameters.

- Once the data is read into the DataFrame, we close the database connection using the `close()` method.

- Finally, we print the first few rows of the DataFrame to verify that the data has been successfully loaded.



**Conclusion:**



Reading data from a SQL database into a Pandas DataFrame is straightforward using the `read_sql()` function. This tutorial provided a clear example and explanation of how to accomplish this task, including establishing a database connection, defining a SQL query, and reading the query results into a DataFrame. By following the steps outlined in this tutorial, you can efficiently import data from SQL databases into Pandas DataFrames for further analysis and manipulation in Python.
