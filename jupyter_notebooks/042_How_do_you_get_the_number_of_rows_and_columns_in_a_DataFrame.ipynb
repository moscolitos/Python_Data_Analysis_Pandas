# How do you get the number of rows and columns in a DataFrame?

**Question:** How do you get the number of rows and columns in a DataFrame?



---



**Introduction:**



Before diving into data analysis and manipulation, it's essential to understand the structure of your dataset. One fundamental aspect is knowing the dimensions of your DataFrame, which includes the number of rows and columns it contains. In this tutorial, we'll explore how to obtain this information using Pandas, a powerful data manipulation library in Python.



**Getting the Number of Rows and Columns:**



Pandas provides two convenient attributes to retrieve the number of rows and columns in a DataFrame:



1. `shape`: This attribute returns a tuple representing the dimensions of the DataFrame. The first element of the tuple corresponds to the number of rows, while the second element corresponds to the number of columns.



2. `len()`: This built-in Python function can be used to get the length of an object. When applied to a DataFrame, `len()` returns the number of rows in the DataFrame.



**Example:**



Let's demonstrate how to use both methods to obtain the number of rows and columns in the Titanic dataset.

````

import pandas as pd



# Load Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Method 1: Using the shape attribute

num_rows, num_columns = titanic_df.shape

print("Number of rows:", num_rows)

print("Number of columns:", num_columns)



# Method 2: Using the len() function

num_rows = len(titanic_df)

num_columns = len(titanic_df.columns)

print("Number of rows:", num_rows)

print("Number of columns:", num_columns)
````

**Output:**

```

Number of rows: 891

Number of columns: 12

Number of rows: 891

Number of columns: 12

```



**Explanation:**



- We first load the Titanic dataset into a DataFrame `titanic_df` using the `read_csv()` function from Pandas.

- We then use two different methods to obtain the number of rows and columns:

  - Method 1: We use the `shape` attribute, which returns a tuple containing the number of rows and columns. We unpack this tuple into variables `num_rows` and `num_columns`.

  - Method 2: We use the `len()` function twice. First, we apply it to the DataFrame `titanic_df` to get the number of rows (`num_rows`). Then, we apply it to `titanic_df.columns` to get the number of columns (`num_columns`).

- Finally, we print out the number of rows and columns obtained using both methods.



**Conclusion:**



Understanding the dimensions of a DataFrame is crucial for data analysis tasks. In this tutorial, we explored two methods to obtain the number of rows and columns in a DataFrame using Pandas: the `shape` attribute and the `len()` function. Both methods provide a simple and effective way to retrieve this essential information, enabling analysts to better understand the structure of their datasets.
