# How do you filter rows based on a specific condition?

**Question:** How do you filter rows based on a specific condition?



---



**Introduction:**



Filtering rows based on specific conditions is a fundamental operation in data analysis. It allows us to extract subsets of data that meet certain criteria, enabling us to focus on relevant information for further analysis. Pandas provides powerful methods for filtering rows in a DataFrame based on conditions, making it easy to extract the desired subset of data.



**Filtering Rows based on Conditions:**



Pandas offers several methods to filter rows based on specific conditions:



1. **Using Boolean Indexing:**

   - We can create boolean masks that represent the conditions we want to apply to the DataFrame.

   - These boolean masks can then be used to filter rows by passing them inside square brackets `[ ]`.



2. **Using the `query()` Method:**

   - Pandas provides the `query()` method, which allows us to filter rows using SQL-like syntax.

   - We can specify the filtering condition as a string within the `query()` method.



**Example:**



Let's demonstrate how to filter rows based on a specific condition using the Titanic dataset. We'll filter the dataset to extract only the rows where passengers are females and survived.

````

import pandas as pd



# Load Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Filter rows based on condition (female passengers who survived)

filtered_df = titanic_df[(titanic_df['Sex'] == 'female') & (titanic_df['Survived'] == 1)]



# Display the filtered DataFrame

print(filtered_df.head())
````

**Output:**

```

[5 rows x 12 columns]

```



**Explanation:**



- We use boolean indexing to create a boolean mask that represents the condition where passengers are females (`'Sex' == 'female'`) and survived (`'Survived' == 1`).

- We then apply this boolean mask to the DataFrame using square brackets `[ ]` to filter the rows that satisfy the condition.

- The resulting `filtered_df` contains only the rows where the condition is true, i.e., female passengers who survived.



**Conclusion:**



Filtering rows based on specific conditions is a crucial skill in data analysis, and Pandas provides efficient methods to accomplish this task. By using boolean indexing or the `query()` method, we can easily extract subsets of data that meet our criteria.
