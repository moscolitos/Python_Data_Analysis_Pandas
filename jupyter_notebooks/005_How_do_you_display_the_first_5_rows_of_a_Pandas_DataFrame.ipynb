# How do you display the first 5 rows of a Pandas DataFrame?

**Question:** How do you display the first 5 rows of a Pandas DataFrame?



---



**Introduction:**



When working with large datasets, it's essential to preview the data to get a glimpse of its structure and contents. Pandas offers convenient methods to view the data within a DataFrame. In this tutorial, we'll explore how to display the first 5 rows of a Pandas DataFrame using simple yet powerful functions.



**Displaying the First 5 Rows of a DataFrame:**



Pandas provides the `DataFrame.head()` method to display the first few rows of a DataFrame. By default, it shows the first 5 rows, but you can specify a different number of rows if needed.



**Example:**



Let's demonstrate how to display the first 5 rows of the Titanic dataset using Pandas:

```

import pandas as pd



# URL of the Titanic dataset CSV file

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"



# Reading CSV file into DataFrame

titanic_df = pd.read_csv(url)



# Displaying the first 5 rows of the DataFrame

print("First 5 rows of the Titanic dataset:")

print(titanic_df.head())
```

In the above code:

- We imported pandas as `pd`, following the conventional alias.

- Specified the URL of the Titanic dataset CSV file.

- Used the `pd.read_csv()` function to read the CSV file from the provided URL and create a DataFrame `titanic_df`.

- Employed the `head()` method of the DataFrame to display the first 5 rows of the dataset.



**Parameters of `head()`:**



The `head()` method provides an optional parameter `n` to specify the number of rows to display. By default, it displays the first 5 rows.



**Conclusion:**



Displaying the first few rows of a Pandas DataFrame is a fundamental operation in data analysis and exploration. Pandas' `head()` method offers a simple yet effective way to preview the data quickly. Understanding this method enables data analysts and scientists to gain insights into the dataset's structure and contents efficiently.
