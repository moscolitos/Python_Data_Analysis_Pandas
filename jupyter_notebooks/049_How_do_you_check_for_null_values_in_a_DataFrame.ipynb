# How do you check for null values in a DataFrame?

**Question:** How do you check for null values in a DataFrame?



---



**Introduction:**



Before proceeding with data analysis or modeling tasks, it's crucial to ensure that the dataset is clean and does not contain missing values. Pandas provides convenient methods for detecting and handling missing data, one of which is checking for null values in a DataFrame. In this tutorial, we will explore how to use Pandas to identify null values within a DataFrame.



**Checking for Null Values in a DataFrame with Pandas:**



Pandas offers the `isnull()` and `isna()` functions to identify missing values in a DataFrame. These functions return a DataFrame of the same shape as the input DataFrame, with boolean values indicating the presence of null values. Additionally, Pandas provides the `info()` method to obtain a concise summary of the DataFrame, including the count of non-null values in each column.



**Example:**



Let's demonstrate how to check for null values in the Titanic dataset using Pandas.

```

import pandas as pd



# Load the Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

df = pd.read_csv(url)



# Check for null values using isnull() or isna() function

null_values = df.isnull()



# Display the count of null values for each column

print(null_values.sum())



# Alternatively, use info() method to get summary information

print(df.info())
```

**Explanation:**



- We begin by importing the Pandas library with the alias `pd`.

- Next, we load the Titanic dataset into a DataFrame `df` using the `read_csv()` function from Pandas. We specify the URL of the dataset as the input.

- To check for null values in the DataFrame, we use the `isnull()` function, which returns a DataFrame of boolean values indicating the presence of null values.

- We can also use the `isna()` function, which is an alias for `isnull()`, providing the same functionality.

- The `sum()` method is then applied to the boolean DataFrame to calculate the count of null values for each column.

- Alternatively, we can use the `info()` method, which provides a concise summary of the DataFrame, including the count of non-null values in each column.



**Conclusion:**



By utilizing the `isnull()` or `isna()` function in Pandas, users can effectively identify missing values within a DataFrame, enabling them to take appropriate actions such as imputation or removal of missing data. Furthermore, the `info()` method provides a quick overview of the dataset's null values, aiding in data exploration and preprocessing tasks.
