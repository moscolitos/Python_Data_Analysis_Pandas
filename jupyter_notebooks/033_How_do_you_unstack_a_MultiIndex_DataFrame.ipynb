# How do you unstack a MultiIndex DataFrame?

**Question:** How do you unstack a MultiIndex DataFrame?



---



**Introduction:**



MultiIndex, also known as hierarchical indexing, is a powerful feature in Pandas that allows you to work with higher-dimensional data in a two-dimensional DataFrame. It enables you to represent and manipulate data with multiple levels of indexing along both rows and columns. However, when dealing with MultiIndex DataFrames, you may often need to restructure or reshape the data for better analysis and visualization. One common operation is unstacking the MultiIndex DataFrame, which involves moving one or more levels of the index to the columns. In this tutorial, we'll explore how to unstack a MultiIndex DataFrame in Pandas using the `unstack` function.



**Unstacking a MultiIndex DataFrame:**



The `unstack` function in Pandas reshapes a MultiIndex DataFrame by pivoting a level of the index labels to the columns. It is particularly useful when you want to transform hierarchical index labels into columns, making the DataFrame more accessible for analysis or visualization.



Syntax:

````

DataFrame.unstack(level=-1, fill_value=None)
````

- `level`: The level of the index labels to unstack. By default, it unstacks the innermost level.

- `fill_value`: Replace missing values with this value.



**Example:**



Let's demonstrate how to unstack a MultiIndex DataFrame using a sample dataset. Suppose we have a DataFrame with a MultiIndex consisting of two levels: 'Gender' and 'Age Group'.

````

import pandas as pd



# Sample DataFrame with MultiIndex

data = {'Score': [85, 90, 88, 92],

        'Grade': ['A', 'A', 'B', 'A']}

index = pd.MultiIndex.from_tuples([('Male', 'Young'), ('Male', 'Old'), ('Female', 'Young'), ('Female', 'Old')], names=['Gender', 'Age Group'])

df = pd.DataFrame(data, index=index)



print("Original DataFrame:")

print(df)



# Unstacking the MultiIndex DataFrame

unstacked_df = df.unstack()

print("\nUnstacked DataFrame:")

print(unstacked_df)
````

**Output:**

```

       Old         90     A

       Old         92     A



Unstacked DataFrame:

         Score       Grade

Gender

Female      92    88     A     B

```



**Explanation:**



- In the example above, we start with a DataFrame `df` with a MultiIndex consisting of two levels: 'Gender' and 'Age Group'.

- We then use the `unstack` function to unstack the DataFrame, which moves the innermost level ('Age Group') of the index labels to the columns, resulting in a reshaped DataFrame `unstacked_df`.

- The resulting DataFrame `unstacked_df` has a single level of column labels, with each unique value of the unstacked index level becoming a separate column.



**Conclusion:**



In this tutorial, we learned how to unstack a MultiIndex DataFrame in Pandas using the `unstack` function. Unstacking a MultiIndex DataFrame is a useful operation for reshaping hierarchical data, making it more accessible for analysis and visualization tasks. By mastering this technique, data analysts can efficiently work with complex, higher-dimensional datasets and extract valuable insights for their projects.
