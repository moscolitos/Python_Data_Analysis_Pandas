# How do you round off numerical values in a DataFrame?

**Question:** How do you round off numerical values in a DataFrame?



---



**Introduction:**



In data analysis, it's often necessary to round off numerical values to a certain number of decimal places for better presentation or analysis. Pandas provides a convenient method to round off numerical values in a DataFrame, allowing analysts to control the precision of the data displayed.



**Rounding Off Numerical Values:**



Pandas offers the `round()` method to round off numerical values in a DataFrame.



- **Syntax:** `DataFrame.round(decimals=)`



  - `DataFrame`: The DataFrame containing the numerical values to be rounded.

  - `decimals`: An integer indicating the number of decimal places to round to. If not specified, it defaults to 0.



**Example:**



Let's round off the 'Fare' column in the Titanic dataset to two decimal places.

````

import pandas as pd



# Load Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Round off the 'Fare' column to two decimal places

titanic_df['Fare'] = titanic_df['Fare'].round(decimals=2)



# Display the updated DataFrame

print(titanic_df.head())
````

**Output:**

```

3            4         1       1  ...    53.10  C123         S

4            5         0       3  ...     8.05   NaN         S



[5 rows x 12 columns]

```



**Explanation:**



- We use the `round()` method on the 'Fare' column of the Titanic DataFrame (`titanic_df['Fare']`) to round off the numerical values to two decimal places.

- The `decimals` parameter is set to 2 to specify rounding off to two decimal places.

- The 'Fare' column in the DataFrame is updated with the rounded values.

- We then display the first five rows of the updated DataFrame using the `head()` method.



**Conclusion:**



Rounding off numerical values in a DataFrame is a common operation in data analysis to control the precision of the data displayed. In this tutorial, we demonstrated how to use the `round()` method in Pandas to round off numerical values in a DataFrame, using the Titanic dataset as an example. By specifying the desired number of decimal places, analysts can effectively manage the precision of their data for better presentation and analysis.
