# How can you perform pivot operations in a DataFrame?

**Question:** How can you perform pivot operations in a DataFrame?



---



**Introduction:**



Pivoting is a crucial operation in data analysis, especially when dealing with tabular data. It involves reshaping the structure of a DataFrame by reorganizing its rows and columns. Pandas, a powerful data manipulation library in Python, provides robust functionalities for performing pivot operations efficiently. In this tutorial, we'll explore how to perform pivot operations in a DataFrame using Pandas, including the `pivot`, `pivot_table`, and `melt` functions.



**Performing Pivot Operations in a DataFrame:**



1. **`pivot` Function:**

   The `pivot` function reshapes the DataFrame by specifying columns to use as new index and columns. It is suitable for pivoting when there are no duplicate entries in the specified index-column pairs.



   Syntax:

```

   DataFrame.pivot(index=None, columns=None, values=None)
```

- `index`: Column to use as the new index.

   - `columns`: Column to use as the new columns.

   - `values`: Column(s) to fill DataFrame values.



**Example:**

```

import pandas as pd



# Sample DataFrame

data = {'Date': ['2022-01-01', '2022-01-01', '2022-01-02', '2022-01-02'],

        'City': ['New York', 'Los Angeles', 'New York', 'Los Angeles'],

        'Temperature': [32, 75, 30, 72],

        'Humidity': [60, 40, 65, 45]}

df = pd.DataFrame(data)



# Pivot DataFrame

pivot_df = df.pivot(index='Date', columns='City', values='Temperature')

print(pivot_df)
```

2. **`pivot_table` Function:**

   The `pivot_table` function creates a pivot table from the DataFrame, allowing aggregation of values with the same index/column pairs. It is useful when there are duplicate entries, and aggregation is needed.



   Syntax:

```

   DataFrame.pivot_table(values=None, index=None, columns=None, aggfunc='mean')
```

- `values`: Column(s) to aggregate.

   - `index`: Column(s) to use as the new index.

   - `columns`: Column(s) to use as the new columns.

   - `aggfunc`: Aggregation function to apply.



**Example:**

```

# Pivot table with aggregation

pivot_table = df.pivot_table(values='Temperature', index='Date', columns='City', aggfunc='mean')

print(pivot_table)
```

3. **`melt` Function:**

   The `melt` function unpivots a DataFrame from wide format to long format, making it easier to perform analysis and visualization. It melts or unpivots a DataFrame from a wide format to a long format.



   Syntax:

```

   DataFrame.melt(id_vars=None, value_vars=None, var_name=None, value_name='value')
```

- `id_vars`: Columns to use as identifier variables.

   - `value_vars`: Columns to unpivot.

   - `var_name`: Name for the variable column.

   - `value_name`: Name for the value column.



**Example:**

```

# Melt DataFrame

melted_df = pivot_df.melt(id_vars='Date', value_vars=['New York', 'Los Angeles'], var_name='City', value_name='Temperature')

print(melted_df)
```

By mastering these pivot operations, data analysts can efficiently reshape and manipulate tabular data to extract valuable insights for analysis and visualization tasks.
