# How do you filter a DataFrame based on a date range?

**Question:** How do you filter a DataFrame based on a date range?



---



**Introduction:**



Filtering a DataFrame based on a date range is a common operation in data analysis, especially when working with time-series data. Pandas, a popular data manipulation library in Python, offers powerful methods for filtering DataFrames based on date and time values. By specifying a date range, analysts can extract relevant subsets of data for further analysis or visualization. In this tutorial, we'll explore how to filter a DataFrame based on a date range using Pandas, providing detailed explanations and coding examples.



**Filtering a DataFrame Based on a Date Range:**



Pandas provides several methods to filter a DataFrame based on a date range:

1. **Using boolean indexing:** This method involves creating boolean masks based on date conditions and applying them to the DataFrame.

2. **Using `.loc[]` accessor:** This method allows selecting rows based on labels and conditions, including date conditions.

3. **Using `.query()` method:** This method enables filtering DataFrame rows using a query expression, including date conditions.



**Example:**



Let's demonstrate how to filter the Titanic dataset DataFrame based on a date range for the 'Embarked' column:

```

import pandas as pd



# URL of the Titanic dataset CSV file

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"



# Reading CSV file into DataFrame

titanic_df = pd.read_csv(url)



# Converting 'Embarked' column to datetime format

titanic_df['Embarked'] = pd.to_datetime(titanic_df['Embarked'])



# Filtering DataFrame based on date range for 'Embarked' column

start_date = '1912-04-10'

end_date = '1912-04-15'

filtered_df = titanic_df[(titanic_df['Embarked'] >= start_date) & (titanic_df['Embarked'] <= end_date)]



# Displaying filtered DataFrame

print("Filtered DataFrame based on date range for 'Embarked' column:")

print(filtered_df)
```

In the above code:

- We imported pandas as `pd`, following the conventional alias.

- Specified the URL of the Titanic dataset CSV file.

- Used the `pd.read_csv()` function to read the CSV file from the provided URL and create a DataFrame `titanic_df`.

- Converted the 'Embarked' column to datetime format using the `pd.to_datetime()` function.

- Defined a start date and end date for the date range filtering.

- Filtered the DataFrame based on the date range for the 'Embarked' column using boolean indexing.

- Displayed the filtered DataFrame containing rows with 'Embarked' dates within the specified range.



**Conclusion:**

By leveraging these methods, analysts can efficiently handle time-series data and derive meaningful insights from their datasets.
