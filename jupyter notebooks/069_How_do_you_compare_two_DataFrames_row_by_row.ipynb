# How do you compare two DataFrames row by row?

**Question:** How do you compare two DataFrames row by row?



---



**Introduction:**



Comparing two DataFrames row by row is a common task in data analysis. It allows you to identify differences or similarities between corresponding rows in two different datasets. Pandas provides several methods to compare DataFrames row by row, such as the `equals()` function, iterating through rows, or using boolean indexing.



**Comparing DataFrames Row by Row:**



To compare two DataFrames row by row, you can use the `equals()` function in Pandas. This function checks whether two DataFrames are equal element-wise, which includes comparing corresponding rows. Additionally, you can iterate through rows using iteration methods like `iterrows()` or `apply()` to perform custom row-wise comparisons.



**Example:**



Let's demonstrate how to compare two DataFrames row by row using the Titanic dataset. In this example, we'll create two identical DataFrames and then compare them row by row using the `equals()` function.

```

import pandas as pd



# Load the Titanic dataset into a DataFrame

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df1 = pd.read_csv(url)

titanic_df2 = pd.read_csv(url)  # Create a copy of the DataFrame



# Compare DataFrames row by row

are_equal = titanic_df1.equals(titanic_df2)



if are_equal:

    print("The two DataFrames are equal row by row.")

else:

    print("The two DataFrames are not equal row by row.")
```

**Explanation:**



- We start by importing the Pandas library with the alias `pd`.

- Next, we load the Titanic dataset from the provided URL into two separate DataFrames, `titanic_df1` and `titanic_df2`, representing the two datasets to be compared.

- To compare the DataFrames row by row, we use the `equals()` function on `titanic_df1` and pass `titanic_df2` as an argument. This function returns `True` if the two DataFrames are equal element-wise, including corresponding rows.

- Finally, we check the result of the comparison and print whether the two DataFrames are equal row by row or not.



**Conclusion:**



Comparing two DataFrames row by row is a useful technique in data analysis to identify similarities or differences between corresponding rows in different datasets. By using the `equals()` function in Pandas, you can efficiently perform this comparison and determine whether the two DataFrames are equal row by row. This allows you to validate data integrity, identify discrepancies, and gain insights into the consistency of datasets.
