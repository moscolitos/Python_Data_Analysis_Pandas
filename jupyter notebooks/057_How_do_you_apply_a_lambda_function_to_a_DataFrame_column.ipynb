# How do you apply a lambda function to a DataFrame column?

**Question:** How do you apply a lambda function to a DataFrame column?



---



**Introduction:**



In pandas, lambda functions provide a convenient way to perform quick transformations on DataFrame columns. A lambda function is an anonymous function that can take any number of arguments but can only have one expression. When applied to DataFrame columns, lambda functions enable users to perform element-wise operations efficiently without the need to define a separate named function.



**Applying a Lambda Function to a DataFrame Column:**



To apply a lambda function to a DataFrame column, you can use the `.apply()` method. This method allows you to apply a function, including lambda functions, along an axis of the DataFrame. By specifying the lambda function as the argument to the `.apply()` method, you can transform the values in the column according to the defined expression.



**Example:**



Let's demonstrate how to apply a lambda function to a DataFrame column using the Titanic dataset. Suppose we want to calculate the square of the 'Fare' values and create a new column called 'Fare_Squared'.

```

import pandas as pd



# Load the Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Apply a lambda function to calculate the square of 'Fare'

titanic_df['Fare_Squared'] = titanic_df['Fare'].apply(lambda x: x**2)



# Display the DataFrame with the new column

print(titanic_df[['Fare', 'Fare_Squared']].head())
```

**Explanation:**



- We begin by importing the pandas library with the alias `pd`.

- Next, we load the Titanic dataset from the provided URL using the `read_csv()` function and store it in the DataFrame `titanic_df`.

- To apply a lambda function to the 'Fare' column, we use the `.apply()` method on the column and pass a lambda function as the argument. The lambda function calculates the square of each value in the 'Fare' column.

- The result of applying the lambda function is assigned to a new column called 'Fare_Squared' in the DataFrame.

- Finally, we display the 'Fare' and 'Fare_Squared' columns of the DataFrame using the `print()` function.



**Conclusion:**



Applying a lambda function to a DataFrame column in pandas allows users to perform element-wise operations efficiently without defining a separate named function. By using the `.apply()` method and specifying a lambda function as the argument, users can easily transform column values according to the defined expression. Whether you need to perform simple arithmetic calculations or more complex transformations, lambda functions provide a flexible and convenient way to manipulate DataFrame columns in pandas.
