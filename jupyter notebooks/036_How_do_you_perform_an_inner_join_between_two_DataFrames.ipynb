# How do you perform an inner join between two DataFrames?

**Question:** How do you perform an inner join between two DataFrames?



---



**Introduction:**



In data analysis, joining datasets is a common operation to combine information from multiple sources. One of the most frequently used types of joins is the "inner join," where only the matching rows between two DataFrames are retained. In this tutorial, we'll explore how to perform an inner join between two DataFrames in Pandas using the `merge()` function.



**Performing an Inner Join:**



The `merge()` function in Pandas is used to combine two DataFrames based on one or more keys. By default, it performs an inner join, meaning it retains only the rows where the key is present in both DataFrames.



Syntax:

````

pd.merge(left, right, how='inner', on=None, left_on=None, right_on=None, left_index=False, right_index=False, sort=False, suffixes=('_x', '_y'), copy=True, indicator=False, validate=None)
````

- `left`: The left DataFrame to be merged.

- `right`: The right DataFrame to be merged.

- `how`: The type of join to be performed. It can be 'left', 'right', 'outer', or 'inner'. Default is 'inner'.

- `on`: Column or index level names to join on. These must be found in both DataFrames. If not specified and no other join keys are given, it will use the intersection of the columns in both DataFrames.

- `left_on`, `right_on`: Columns or index levels from the left and right DataFrames to use as keys.

- `left_index`, `right_index`: If True, use the index from the left/right DataFrame as the join key(s).



**Example:**



Let's demonstrate how to perform an inner join between two DataFrames using a sample dataset. We'll start with two DataFrames and then perform an inner join based on a common key.

````

import pandas as pd



# Sample DataFrames

left_df = pd.DataFrame({'key': ['A', 'B', 'C', 'D'],

                        'value_left': [1, 2, 3, 4]})

right_df = pd.DataFrame({'key': ['B', 'D', 'E', 'F'],

                         'value_right': [5, 6, 7, 8]})



print("Left DataFrame:")

print(left_df)

print("\nRight DataFrame:")

print(right_df)



# Performing an inner join

result_df = pd.merge(left_df, right_df, how='inner', on='key')



print("\nInner Join Result:")

print(result_df)
````

**Output:**

```

Left DataFrame:

  key  value_left

0   A           1

1   B           2

2   C           3

3   D           4



Right DataFrame:

  key  value_right

0   B            5

1   D            6

2   E            7

3   F            8



Inner Join Result:

  key  value_left  value_right

0   B           2            5

1   D           4            6

```



**Explanation:**



- In the example above, we have two sample DataFrames `left_df` and `right_df`, each containing a 'key' column.

- We use the `merge()` function to perform an inner join on the 'key' column, retaining only the rows where the key is present in both DataFrames.

- The resulting DataFrame `result_df` contains the merged data, including only the rows with matching keys from both DataFrames.



**Conclusion:**



In this tutorial, we learned how to perform an inner join between two DataFrames in Pandas using the `merge()` function. Inner joins are useful for combining datasets while retaining only the rows that have matching keys in both DataFrames. By understanding the syntax and parameters of the `merge()` function, data analysts can efficiently merge datasets to consolidate and analyze information from multiple sources.
