# How do you plot a line graph from a DataFrame column?

**Question:** How do you plot a line graph from a DataFrame column?



---



**Introduction:**



Line graphs are commonly used to visualize trends over time or relationships between variables. In Pandas, you can easily create a line graph from a DataFrame column using the `plot()` function with the `kind='line'` parameter.



**Plotting a Line Graph:**



To plot a line graph from a DataFrame column in Pandas, you can use the `plot()` function with the `kind='line'` parameter. This function generates a line graph where the x-axis represents the index or another column of the DataFrame, and the y-axis represents the values of the specified column.



**Example:**



Let's demonstrate how to plot a line graph from the 'Age' column of the Titanic dataset. In this example, we'll create a line graph to visualize the distribution of passenger ages.

```

import pandas as pd

import matplotlib.pyplot as plt



# Load the Titanic dataset into a DataFrame

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Plot a line graph of passenger ages

titanic_df['Age'].plot(kind='line')



# Set plot labels and title

plt.xlabel('Passenger Index')

plt.ylabel('Age')

plt.title('Distribution of Passenger Ages')



# Show the plot

plt.show()
```

**Explanation:**



- We start by importing the Pandas library with the alias `pd` and the Matplotlib library with the alias `plt`.

- Next, we load the Titanic dataset from the provided URL into a DataFrame named `titanic_df`.

- Using the `plot()` function with `kind='line'`, we create a line graph of the 'Age' column from the Titanic dataset. By default, the index of the DataFrame is used as the x-axis values.

- We customize the plot by setting labels for the x-axis and y-axis using the `xlabel()` and `ylabel()` functions, respectively. Additionally, we add a title to the plot using the `title()` function.

- Finally, we display the plot using the `show()` function.



**Conclusion:**



Plotting a line graph from a DataFrame column in Pandas allows you to visualize trends or patterns in your data. By following the steps outlined in this tutorial, you can easily generate informative line graphs to analyze the distribution of values within a specific column of your dataset.
