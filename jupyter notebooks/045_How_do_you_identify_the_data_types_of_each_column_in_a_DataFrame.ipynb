# How do you identify the data types of each column in a DataFrame?

**Question:** How do you identify the data types of each column in a DataFrame?



---



**Introduction:**



Before performing any data analysis or manipulation, it's essential to understand the structure of your dataset, including the data types of each column. Pandas provides a convenient method to inspect the data types of DataFrame columns, allowing you to ensure that the data is correctly interpreted and processed according to your analysis requirements. In this tutorial, we'll explore how to identify the data types of each column in a DataFrame using Pandas.



**Identifying Data Types:**



Pandas DataFrame objects have a `dtypes` attribute that allows you to retrieve the data types of each column in the DataFrame. This attribute returns a Pandas Series containing the data types of each column, where the index represents the column names, and the values represent the corresponding data types.



**Example:**



Let's demonstrate how to use the `dtypes` attribute to identify the data types of each column in a DataFrame.

```

import pandas as pd



# Load the Titanic dataset into a DataFrame

url = "https://github.com/moscolitos/titanic-dataset/raw/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Display the data types of each column

print(titanic_df.dtypes)
```

**Explanation:**



- We start by importing the Pandas library with the alias `pd`.

- Next, we load the Titanic dataset into a DataFrame `titanic_df` using the `read_csv()` function from Pandas. We pass the URL of the dataset as a parameter.

- To identify the data types of each column, we access the `dtypes` attribute of the DataFrame `titanic_df`. This attribute returns a Pandas Series containing the data types of each column.

- Finally, we print the Pandas Series containing the data types, which displays the data type of each column along with the corresponding column names.



**Conclusion:**



Identifying the data types of each column in a DataFrame is crucial for understanding the structure of your dataset and ensuring that data is correctly interpreted during analysis. This tutorial provided a clear example and explanation of how to use the `dtypes` attribute in Pandas to retrieve the data types of DataFrame columns. By following the steps outlined in this tutorial, you can easily inspect the data types of each column in your DataFrame, allowing for more effective data analysis and manipulation in Python.
