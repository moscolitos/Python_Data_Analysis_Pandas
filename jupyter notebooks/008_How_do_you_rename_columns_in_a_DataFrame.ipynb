# How do you rename columns in a DataFrame?

**Question:** How do you rename columns in a DataFrame?



---



**Introduction:**



Column names play a crucial role in identifying and accessing data within a DataFrame. Sometimes, you may need to rename columns to make them more descriptive or align them with specific conventions. Pandas provides convenient methods to rename columns in a DataFrame, enabling data analysts to customize column names as needed. In this tutorial, we'll explore how to rename columns in a Pandas DataFrame with detailed explanations and coding examples.



**Renaming Columns in a DataFrame:**



Pandas offers the `DataFrame.rename()` method to rename columns in a DataFrame. This method allows analysts to specify new names for one or more columns and customize the renaming process according to specific requirements.



**Example:**



Let's demonstrate how to rename columns in the Titanic dataset to make them more descriptive:

```

import pandas as pd



# URL of the Titanic dataset CSV file

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"



# Reading CSV file into DataFrame

titanic_df = pd.read_csv(url)



# Renaming columns in the DataFrame

titanic_df = titanic_df.rename(columns={

    'Pclass': 'Passenger_Class',

    'SibSp': 'Siblings_Spouses',

    'Parch': 'Parents_Children',

    'Fare': 'Ticket_Fare',

    'Embarked': 'Port_of_Embarkation'

})



# Displaying the first few rows of the DataFrame with renamed columns

print("DataFrame with renamed columns:")

print(titanic_df.head())
```

In the above code:

- We imported pandas as `pd`, following the conventional alias.

- Specified the URL of the Titanic dataset CSV file.

- Used the `pd.read_csv()` function to read the CSV file from the provided URL and create a DataFrame `titanic_df`.

- Applied the `DataFrame.rename()` method to rename columns in the DataFrame. We provided a dictionary where keys represent the original column names and values represent the new column names.

- Displayed the first few rows of the DataFrame with renamed columns.



**Parameters of `DataFrame.rename()`:**



The `DataFrame.rename()` method allows for renaming one or more columns in a DataFrame. Some commonly used parameters include:

- **columns**: Specifies a dictionary mapping old column names to new column names.

- **inplace**: Specifies whether to modify the DataFrame in place or return a new DataFrame with the renamed columns. By default, it is False.



**Conclusion:**



Renaming columns in a Pandas DataFrame is a crucial operation for data preprocessing and customization. Pandas' `DataFrame.rename()` method provides a flexible and efficient way to rename columns based on specific requirements. Mastering this method empowers data analysts and scientists to manipulate and organize data effectively.
