# How can you add a new column to a DataFrame?

**Question:** How can you add a new column to a DataFrame?



---



**Introduction:**



Enhancing a DataFrame by adding new columns is a common task in data analysis. Pandas provides convenient functionalities to append additional information to a DataFrame, enabling data analysts to enrich their datasets with derived or calculated values. In this tutorial, we'll explore how to add a new column to a Pandas DataFrame with detailed explanations and coding examples.



**Adding a New Column to a DataFrame:**



Pandas offers multiple approaches to add a new column to a DataFrame. One common method is using the bracket notation (`[]`) to assign values to a new column. Alternatively, you can utilize the `DataFrame.assign()` method to add a column based on existing data or calculations.



**Example:**



Let's demonstrate how to add a new column 'Family_Size' to the Titanic dataset, representing the total number of family members (including the passenger) onboard:

```

import pandas as pd



# URL of the Titanic dataset CSV file

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"



# Reading CSV file into DataFrame

titanic_df = pd.read_csv(url)



# Adding a new column 'Family_Size' to the DataFrame

titanic_df['Family_Size'] = titanic_df['SibSp'] + titanic_df['Parch'] + 1



# Displaying the first few rows of the DataFrame with the new column

print("DataFrame with the new column 'Family_Size':")

print(titanic_df.head())
```

In the above code:

- We imported pandas as `pd`, following the conventional alias.

- Specified the URL of the Titanic dataset CSV file.

- Used the `pd.read_csv()` function to read the CSV file from the provided URL and create a DataFrame `titanic_df`.

- Added a new column 'Family_Size' to the DataFrame using the bracket notation (`[]`). We calculated the total family size by summing the 'SibSp' (number of siblings/spouses) and 'Parch' (number of parents/children) columns, and added 1 to include the passenger themselves.

- Displayed the first few rows of the DataFrame with the new column 'Family_Size'.



**Conclusion:**

Mastering these techniques empowers data analysts and scientists to augment and customize datasets according to specific analytical needs.
