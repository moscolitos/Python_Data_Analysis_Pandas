# How do you create a cross-tabulation from a DataFrame?

**Question:** How do you create a cross-tabulation from a DataFrame?



---



**Introduction:**



Cross-tabulation, also known as contingency table or crosstab, is a useful way to summarize and analyze the relationship between two or more categorical variables. It displays the frequency distribution of variables in a tabular format, making it easier to understand the relationship between them. In Pandas, you can create a cross-tabulation from a DataFrame using the `pd.crosstab()` function.



**Creating a Cross-Tabulation:**



To create a cross-tabulation from a DataFrame in Pandas, you can use the `pd.crosstab()` function. This function takes one or more DataFrame columns as inputs and generates a cross-tabulation table based on their values. Additionally, you can specify row and column margins to include row and column totals in the resulting table.



**Example:**



Let's demonstrate how to create a cross-tabulation from the Titanic dataset. In this example, we'll create a cross-tabulation table to analyze the relationship between the 'Pclass' (Passenger Class) and 'Survived' columns.

```

import pandas as pd



# Load the Titanic dataset into a DataFrame

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Create a cross-tabulation of 'Pclass' and 'Survived' columns

cross_tab = pd.crosstab(titanic_df['Pclass'], titanic_df['Survived'], margins=True)



print(cross_tab)
```

**Explanation:**



- We start by importing the Pandas library with the alias `pd`.

- Next, we load the Titanic dataset from the provided URL into a DataFrame named `titanic_df`.

- To create the cross-tabulation, we use the `pd.crosstab()` function and pass the 'Pclass' column as the first argument and the 'Survived' column as the second argument. This generates a cross-tabulation table that shows the frequency distribution of passengers based on their passenger class and survival status.

- We also set the `margins` parameter to `True` to include row and column totals in the resulting table.

- Finally, we print the cross-tabulation table to display the relationship between passenger class and survival.



**Conclusion:**



Creating a cross-tabulation from a DataFrame in Pandas provides valuable insights into the relationship between categorical variables. By using the `pd.crosstab()` function, you can summarize and analyze the distribution of variables in a tabular format, making it easier to identify patterns and trends in the data. This allows for a deeper understanding of the underlying relationships between different categorical variables, facilitating more informed decision-making in data analysis and interpretation.
