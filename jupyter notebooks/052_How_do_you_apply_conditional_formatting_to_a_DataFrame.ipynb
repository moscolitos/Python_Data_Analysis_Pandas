# How do you apply conditional formatting to a DataFrame?

**Question:** How do you apply conditional formatting to a DataFrame?



---



**Introduction:**



Conditional formatting is a powerful feature that allows you to visually highlight certain cells or rows in a DataFrame based on specific conditions. This technique is commonly used in data analysis and visualization to emphasize important insights or patterns in the data. Pandas provides functionalities to apply conditional formatting directly to DataFrames, making it easier to customize the appearance of data based on predefined criteria.



**Applying Conditional Formatting with Pandas:**



To apply conditional formatting to a DataFrame in Pandas, you can use the `style` attribute along with the `applymap()` function. The `applymap()` function enables you to apply a function to each element of the DataFrame, and the returned values determine the style of the cells. By defining custom functions that evaluate conditions, you can specify the formatting rules and visually enhance the DataFrame.



**Example:**



Let's demonstrate how to apply conditional formatting to the Titanic dataset based on the 'Fare' column. We'll highlight cells with fares greater than $50 in green and those less than or equal to $50 in red.

```

import pandas as pd



# Load the Titanic dataset

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"

titanic_df = pd.read_csv(url)



# Define a function to apply conditional formatting

def highlight_fare(val):

    color = 'green' if val > 50 else 'red'

    return f'background-color: {color}'



# Apply conditional formatting to the 'Fare' column

formatted_df = titanic_df.style.applymap(highlight_fare, subset=['Fare'])



# Display the formatted DataFrame

formatted_df
```

**Explanation:**



- We begin by importing the Pandas library with the alias `pd`.

- Next, we load the Titanic dataset from the provided URL using the `read_csv()` function from Pandas and store it in the DataFrame `titanic_df`.

- We define a custom function `highlight_fare()` that takes a value as input and returns a string specifying the background color based on the condition (green if the fare is greater than $50, red otherwise).

- Using the `style` attribute of the DataFrame, we apply the `applymap()` function to map the `highlight_fare()` function to each element of the 'Fare' column. We specify the subset parameter to apply conditional formatting only to the 'Fare' column.

- Finally, we display the formatted DataFrame, which visually highlights cells in green for fares greater than $50 and in red for fares less than or equal to $50.



**Conclusion:**



Conditional formatting is a valuable tool for enhancing data visualization and interpretation in DataFrame analysis. By applying formatting rules based on specific conditions, users can effectively emphasize relevant information and identify patterns in their data. With Pandas, implementing conditional formatting is straightforward, allowing for seamless customization of DataFrame appearance to suit the analytical needs of the user.
