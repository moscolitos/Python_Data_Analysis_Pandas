# How can you get a list of all unique values in a specific column?

**Question:** How can you get a list of all unique values in a specific column?



---



**Introduction:**



When working with large datasets, it's often necessary to examine the unique values present in specific columns. Understanding the unique values in a column is crucial for data exploration, quality assurance, and preprocessing tasks. Pandas, a powerful data manipulation library in Python, offers convenient methods for extracting unique values from DataFrame columns. In this tutorial, we'll explore how to retrieve a list of all unique values in a specific column using Pandas, providing detailed explanations and coding examples.



**Getting Unique Values in a Specific Column:**



Pandas provides the following method to retrieve unique values in a specific column:

- **`unique()` function:** This function returns an array containing the unique values present in the specified column of the DataFrame.



**Example:**



Let's demonstrate how to get a list of all unique values in the 'Embarked' column of the Titanic dataset DataFrame:

```

import pandas as pd



# URL of the Titanic dataset CSV file

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"



# Reading CSV file into DataFrame

titanic_df = pd.read_csv(url)



# Getting a list of all unique values in the 'Embarked' column

unique_values = titanic_df['Embarked'].unique()



# Displaying the list of unique values

print("Unique values in the 'Embarked' column:")

print(unique_values)
```

In the above code:

- We imported pandas as `pd`, following the conventional alias.

- Specified the URL of the Titanic dataset CSV file.

- Used the `pd.read_csv()` function to read the CSV file from the provided URL and create a DataFrame `titanic_df`.

- Retrieved a list of all unique values in the 'Embarked' column using the `unique()` function and assigned it to the `unique_values` variable.

- Displayed the list of unique values in the 'Embarked' column to verify the result.



**Conclusion:**



Retrieving unique values from specific columns is a fundamental operation in data analysis, enabling analysts to gain insights into the diversity and distribution of data. Pandas provides a simple and efficient method, the `unique()` function, for extracting unique values from DataFrame columns. By leveraging this method, analysts can explore and understand the distinct values present in their datasets, facilitating further analysis and decision-making processes.
