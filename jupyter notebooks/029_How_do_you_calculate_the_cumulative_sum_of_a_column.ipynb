# How do you calculate the cumulative sum of a column?

**Question:** How do you calculate the cumulative sum of a column?



---



**Introduction:**



Calculating the cumulative sum of a column in a DataFrame is a common operation in data analysis, especially when dealing with time series or sequential data. The cumulative sum represents the total sum of values up to a certain point in the sequence. Pandas, a popular data manipulation library in Python, provides a simple method to compute the cumulative sum of a column efficiently. In this tutorial, we'll explore how to calculate the cumulative sum of a column in a DataFrame using Pandas, accompanied by detailed explanations and coding examples.



**Calculating Cumulative Sum in Pandas:**



Pandas offers the following method to compute the cumulative sum of a column:

- **`cumsum()` function:** This function computes the cumulative sum of values along a specified axis (default axis is 0, which corresponds to rows) in a DataFrame.



**Example:**



Let's demonstrate how to calculate the cumulative sum of the "Fare" column in the Titanic dataset DataFrame:

```

import pandas as pd



# URL of the Titanic dataset CSV file

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"



# Reading CSV file into DataFrame

titanic_df = pd.read_csv(url)



# Calculating the cumulative sum of the "Fare" column

titanic_df['Cumulative_Fare'] = titanic_df['Fare'].cumsum()



# Displaying the DataFrame with cumulative sum

print("DataFrame with Cumulative Fare:")

print(titanic_df[['Fare', 'Cumulative_Fare']])
```

In the above code:

- We imported pandas as `pd`, following the conventional alias.

- Specified the URL of the Titanic dataset CSV file.

- Used the `pd.read_csv()` function to read the CSV file from the provided URL and create a DataFrame `titanic_df`.

- Applied the `cumsum()` function to compute the cumulative sum of the "Fare" column and assigned the result to a new column "Cumulative_Fare" in the DataFrame.

- Displayed the DataFrame with the original "Fare" column and the newly added "Cumulative_Fare" column.



**Conclusion:**



Calculating the cumulative sum of a column in a DataFrame is a straightforward task in Pandas, thanks to its `cumsum()` function. By computing the cumulative sum, analysts can gain insights into the evolving total of a sequential variable over time or other ordered sequences. This tutorial demonstrates how to leverage Pandas to efficiently calculate the cumulative sum, facilitating insightful data analysis and visualization.
