# How do you apply a function to each row or column of a DataFrame?

**Question:** How do you apply a function to each row or column of a DataFrame?



---



**Introduction:**



In data analysis, it's often necessary to apply custom or built-in functions to each row or column of a DataFrame to perform specific calculations or transformations. Pandas, a powerful data manipulation library in Python, offers flexible methods to apply functions element-wise, row-wise, or column-wise across DataFrames. This capability allows analysts to perform complex operations efficiently and customize data processing according to their requirements. In this tutorial, we'll explore how to apply a function to each row or column of a DataFrame using Pandas, providing detailed explanations and coding examples.



**Applying a Function to Each Row or Column:**



Pandas provides several methods to apply a function to each row or column of a DataFrame:

1. **Using `.apply()`:** This method applies a function along an axis of the DataFrame, either row-wise or column-wise, based on the `axis` parameter.

2. **Using `.applymap()`:** This method applies a function element-wise to the entire DataFrame.

3. **Using `.map()`:** This method applies a function element-wise to a Series.



**Example:**



Let's demonstrate how to apply a function to each row or column of the Titanic dataset DataFrame:

```

import pandas as pd



# URL of the Titanic dataset CSV file

url = "https://raw.githubusercontent.com/moscolitos/titanic-dataset/main/Titanic-Dataset.csv"



# Reading CSV file into DataFrame

titanic_df = pd.read_csv(url)



# Define a custom function to calculate total family size

def calculate_family_size(row):

    family_size = row['SibSp'] + row['Parch'] + 1  # Include the passenger itself

    return family_size



# Apply the custom function to each row to calculate family size

titanic_df['FamilySize'] = titanic_df.apply(calculate_family_size, axis=1)



# Display the DataFrame with the new 'FamilySize' column

print(titanic_df[['Name', 'SibSp', 'Parch', 'FamilySize']].head())
```

In the above code:

- We imported pandas as `pd`, following the conventional alias.

- Specified the URL of the Titanic dataset CSV file.

- Used the `pd.read_csv()` function to read the CSV file from the provided URL and create a DataFrame `titanic_df`.

- Defined a custom function `calculate_family_size(row)` to calculate the total family size based on the 'SibSp' (number of siblings/spouses) and 'Parch' (number of parents/children) columns for each passenger.

- Applied the custom function to each row of the DataFrame using the `.apply()` method with `axis=1` to indicate row-wise operation.

- Created a new column 'FamilySize' in the DataFrame to store the calculated family size.

- Displayed the DataFrame with the 'Name', 'SibSp', 'Parch', and 'FamilySize' columns using the `.head()` method to show the first few rows.



**Conclusion:**



Applying a function to each row or column of a DataFrame is a powerful technique in data analysis, enabling analysts to perform customized calculations or transformations efficiently. Pandas' `.apply()` method, along with other related methods like `.applymap()` and `.map()`, provides flexible options for applying functions to DataFrames, offering versatility and convenience in data processing tasks.
